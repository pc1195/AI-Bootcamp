```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

text = """
My hobbies and interests reflect my passion for learning, exploring, and creating. I enjoy playing cricket, listening to music, travelling, exploring mountains, driving cars, spending time with friends, and discovering new things. I’m passionate about software development, AI & ML, coding, content creation, business, and entrepreneurship. I also enjoy bhakti and spiritual activities that keep me positive and focused. I love building new ideas, exploring opportunities, and continuously learning to grow both personally and professionally.

"""

# Create Word Cloud
wordcloud = WordCloud(
    width=1000,
    height=600,
    background_color="white"
).generate(text)

# Display Word Cloud
plt.figure(figsize=(12, 7))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Artificial Intelligence Word Cloud")
plt.show()

# Save as JPG
wordcloud.to_file("AI_WordCloud.jpg")

print("Word cloud saved successfully as AI_WordCloud.jpg")
```

```python
import cv2
import matplotlib.pyplot as plt

# -----------------------------------------
# 1. Read the image
# -----------------------------------------
img = cv2.imread("image1.jpeg")

if img is None:
    print("Error: Image not found!")
    exit()

# OpenCV reads image in BGR format
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


# -----------------------------------------
# 2. Resize
# -----------------------------------------
resized = cv2.resize(img, (500, 350))


# -----------------------------------------
# 3. Crop
# -----------------------------------------
cropped = img[50:300, 100:450]


# -----------------------------------------
# 4. Flip
# -----------------------------------------
flipped = cv2.flip(img, 1)       # 1 = Horizontal flip


# -----------------------------------------
# 5. Rotate
# -----------------------------------------
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)


# -----------------------------------------
# 6. Convert to Grayscale
# -----------------------------------------
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


# -----------------------------------------
# 7. Blur the image
# -----------------------------------------
blurred = cv2.GaussianBlur(img, (15, 15), 0)


# -----------------------------------------
# 8. Edge Detection
# -----------------------------------------
edges = cv2.Canny(gray, 100, 200)


# -----------------------------------------
# 9. Brightness adjustment
# -----------------------------------------
bright = cv2.convertScaleAbs(img, alpha=1.2, beta=40)


# -----------------------------------------
# 10. Save edited images
# -----------------------------------------
cv2.imwrite("resized.jpg", resized)
cv2.imwrite("cropped.jpg", cropped)
cv2.imwrite("flipped.jpg", flipped)
cv2.imwrite("rotated.jpg", rotated)
cv2.imwrite("grayscale.jpg", gray)
cv2.imwrite("blurred.jpg", blurred)
cv2.imwrite("edges.jpg", edges)
cv2.imwrite("bright.jpg", bright)


# -----------------------------------------
# 11. Display results
# -----------------------------------------
plt.figure(figsize=(15, 10))

plt.subplot(2, 4, 1)
plt.imshow(img_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(2, 4, 2)
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Resized")
plt.axis("off")

plt.subplot(2, 4, 3)
plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))
plt.title("Cropped")
plt.axis("off")

plt.subplot(2, 4, 4)
plt.imshow(cv2.cvtColor(flipped, cv2.COLOR_BGR2RGB))
plt.title("Flipped")
plt.axis("off")

plt.subplot(2, 4, 5)
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title("Rotated")
plt.axis("off")

plt.subplot(2, 4, 6)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 4, 7)
plt.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB))
plt.title("Blurred")
plt.axis("off")

plt.subplot(2, 4, 8)
plt.imshow(edges, cmap="gray")
plt.title("Edge Detection")
plt.axis("off")

plt.tight_layout()
plt.show()

print("All image editing operations completed successfully!")
```
