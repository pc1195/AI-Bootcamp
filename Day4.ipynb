```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("photo.jpg")
img1 =cv2.imread("combine.jpg")

# OpenCV reads images as BGR
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)

#Resize the image
img = cv2.resize(img, (500, 400))
img1 = cv2.resize(img1, (500, 400))

#Or we can use subplot for printing  both the images:
#plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.axis("off")
plt.show()
#plt.subplot(1, 2, 2)
plt.imshow(img1_rgb)
plt.axis("off")
plt.show()
```

```python
img = cv2.imread("photo.jpg")
img1 =cv2.imread("combine.jpg")

# Flip images
img_flip = cv2.flip(img, 1)
img1_flip = cv2.flip(img1, 1)

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img_flip, cv2.COLOR_BGR2RGB)
img1_rgb = cv2.cvtColor(img1_flip, cv2.COLOR_BGR2RGB)

# Display
plt.imshow(img_rgb)
plt.axis("off")
plt.show()
plt.imshow(img1_rgb)
plt.axis("off")
plt.show()
```
