```python
import cv2
import matplotlib.pyplot as plt

# -----------------------------------
# Load Images
# -----------------------------------
img = cv2.imread("photo.jpg")
img1 = cv2.imread("combine.jpg")

# Check if images are loaded
if img is None or img1 is None:
    print("Error: Image not found!")
    exit()

# -----------------------------------
# 1. Flip Images
# -----------------------------------
img_flip = cv2.flip(img, 1)
img1_flip = cv2.flip(img1, 1)

# -----------------------------------
# 2. Resize Images
# -----------------------------------
img_resize = cv2.resize(img_flip, (500, 400))
img1_resize = cv2.resize(img1_flip, (500, 400))

# -----------------------------------
# 3. Convert to Grayscale
# -----------------------------------
gray = cv2.cvtColor(img_resize, cv2.COLOR_BGR2GRAY)
gray1 = cv2.cvtColor(img1_resize, cv2.COLOR_BGR2GRAY)

# -----------------------------------
# 4. Crop Image
# -----------------------------------
crop = img_resize[50:350, 100:400]

# -----------------------------------
# 5. Rotate Image
# -----------------------------------
rotate = cv2.rotate(img_resize, cv2.ROTATE_90_CLOCKWISE)

# -----------------------------------
# 6. Brightness and Contrast
# -----------------------------------
bright = cv2.convertScaleAbs(
    img_resize,
    alpha=1.3,
    beta=40
)

# -----------------------------------
# 7. Blur Image
# -----------------------------------
blur = cv2.GaussianBlur(img_resize, (7, 7), 0)

# -----------------------------------
# 8. Edge Detection
# -----------------------------------
edges = cv2.Canny(gray, 100, 200)

# -----------------------------------
# 9. Draw Rectangle
# -----------------------------------
edited = img_resize.copy()

cv2.rectangle(
    edited,
    (50, 50),
    (300, 300),
    (0, 255, 0),
    3
)

# -----------------------------------
# 10. Draw Circle
# -----------------------------------
cv2.circle(
    edited,
    (400, 300),
    40,
    (255, 0, 0),
    3
)

# -----------------------------------
# 11. Add Text
# -----------------------------------
cv2.putText(
    edited,
    "OpenCV Image Editing",
    (50, 380),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.8,
    (0, 0, 255),
    2
)

# -----------------------------------
# Convert Images BGR -> RGB
# -----------------------------------
img_rgb = cv2.cvtColor(img_resize, cv2.COLOR_BGR2RGB)
img1_rgb = cv2.cvtColor(img1_resize, cv2.COLOR_BGR2RGB)
bright_rgb = cv2.cvtColor(bright, cv2.COLOR_BGR2RGB)
blur_rgb = cv2.cvtColor(blur, cv2.COLOR_BGR2RGB)
edited_rgb = cv2.cvtColor(edited, cv2.COLOR_BGR2RGB)

# -----------------------------------
# Display Images
# -----------------------------------

plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(img_rgb)
plt.title("Original / Resized")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(img1_rgb)
plt.title("Second Image")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(bright_rgb)
plt.title("Brightness Adjusted")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(blur_rgb)
plt.title("Blurred")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(edited_rgb)
plt.title("Final Edited Image")
plt.axis("off")

plt.show()

# -----------------------------------
# Display Edge Detection
# -----------------------------------
plt.imshow(edges, cmap="gray")
plt.title("Edge Detection")
plt.axis("off")
plt.show()

# -----------------------------------
# Save Final Image
# -----------------------------------
cv2.imwrite("final_edited.jpg", edited)

print("Image editing completed successfully!")
```
