```python
# Import necessary libraries
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Text
text = """Basketball has always been one of my favorite hobbies. I enjoy playing basketball in my free time because it keeps me active and teaches me the importance of teamwork, discipline, and patience. I have also spent most of my life living in Air Force stations because my father was an Air Force officer. Growing up in that environment gave me the opportunity to meet people from different places and experience a unique lifestyle. Life in an Air Force station was often disciplined and organized, but it was also full of memorable experiences. I enjoyed participating in sports, making friends, and exploring different activities. These experiences have made me adaptable, confident, and comfortable meeting new people. Apart from basketball, I am interested in technology, learning new skills, and exploring topics related to artificial intelligence and data science. I enjoy learning through practical activities and trying new things whenever I get the opportunity."""

print(text)

# Tokenization
words = word_tokenize(text.lower())

# Print words
print(words)
```

```python
# Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word.isalpha() and word not in stop_words
]

# Join words
clean_text = " ".join(filtered_words)

# Create WordCloud
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(clean_text)

# Display
plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```
