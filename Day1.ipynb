```python
#Create a list of random numbers and print all the even numbers in the list
```

```python
import random

# Create a list of 10 random numbers
numbers = [random.randint(1, 100) for _ in range(10)]

print("Random numbers:", numbers)

# Print even numbers
print("Even numbers:")

for num in numbers:
    if num % 2 == 0:
        print(num)
```

```python
#Create a string and print it in reverse order. Now reverse the words also.
```

```python
# Create a string
text = "Python is very easy"

# Reverse the complete string
reverse_string = text[::-1]
print("Reverse string:", reverse_string)

# Reverse the order of words
reverse_words = " ".join(text.split()[::-1])
print("Reverse words:", reverse_words)
```
